```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
raccoons = pd.read_csv("Toronto Raccoon Activity Index.csv")

print(raccoons)
```

```python
raccoongraph=sns.lineplot(data=raccoons,
                         x='incidents_of_raccoon_human_standoff',
                         y='raccoon_confidence_level',
                         hue = 'complaints_logged',
                         palette = ['#140A0A', '#FAA03C', '#04B46E', '#A08C82', "#6E320A", "#C8B446"],
                         linestyle='-',
                         linewidth=3,
                         marker='*',
                         markersize=12,
                        )

raccoongraph.set(title='The Relationship Between Raccoon Confidence and Raccoon-Human Standoffs in Toronto',
             xlabel='Incidents of Raccoon-Human Standoff',
             ylabel='Raccoon Confidence Level (Out of 10)')

# colour palette derived from https://icolorpalette.com/imagepalette/color-palette-ideas-from-mammal-fur-raccoon-image/ 

```
